```python

average_rating = df["Review Rating"].mean()

print("Average Customer Rating:", round(average_rating, 2))
```

```python
rating_counts = df["Review Rating"].value_counts().sort_index()

print(rating_counts)
```

```python

positive_reviews = (df["Review Rating"] >= 4).sum()
negative_reviews = (df["Review Rating"] <= 2).sum()
neutral_reviews = (df["Review Rating"] == 3).sum()

print("Positive Reviews:", positive_reviews)
print("Negative Reviews:", negative_reviews)
print("Neutral Reviews:", neutral_reviews)
```

```python

total_reviews = len(df)

positive_percentage = (positive_reviews / total_reviews) * 100
negative_percentage = (negative_reviews / total_reviews) * 100
neutral_percentage = (neutral_reviews / total_reviews) * 100

print("Positive Reviews:", round(positive_percentage, 2), "%")
print("Negative Reviews:", round(negative_percentage, 2), "%")
print("Neutral Reviews:", round(neutral_percentage, 2), "%")
```

```python
import pandas as pd

df = pd.read_csv(
    "D:/Desktop/Final Portfolio/Amazon Product Reviews Dataset/amazon_reviews_clean.csv"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
category_review_counts = df["Category"].value_counts()

print(category_review_counts)
```

```python
category_analysis = df["Category"].value_counts().reset_index()

category_analysis.columns = ["Category", "Review Count"]

category_analysis["Percentage"] = (
    category_analysis["Review Count"] / len(df) * 100
)

print(category_analysis.round(2))
```

```python
### Category Analysis — Business Insight

#Beauty & Personal Care accounts for 72% of the reviews in this dataset, while Grocery & Gourmet Food represents 26%. This indicates that the dataset is heavily concentrated in the Beauty & Personal Care category. Therefore, category-level comparisons should be interpreted carefully, as the two categories have different sample sizes.
```

```python
product_review_counts = df["Asin"].value_counts()

print(product_review_counts.head(10))
```

```python
#Product Analysis — Business Insight

#Three products received the highest number of reviews, with 10 reviews each. Another product received 9 reviews. This indicates that customer feedback is concentrated among a small number of products in the dataset. Further analysis of their average ratings and negative reviews could help identify products that require attention.
```

```python
product_analysis = df.groupby("Asin").agg(
    review_count=("Review Rating", "count"),
    average_rating=("Review Rating", "mean")
)

product_analysis = product_analysis.sort_values(
    by="review_count",
    ascending=False
)

print(product_analysis.round(2))
```

```python
verified_purchase_analysis = df.groupby("Verified Purchase").agg(
    review_count=("Review Rating", "count"),
    average_rating=("Review Rating", "mean")
)

print(verified_purchase_analysis.round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "D:/Desktop/Final Portfolio/Amazon Product Reviews Dataset/amazon_reviews_clean.csv"
)
```

```python
brand_ratings = df.groupby("Brand")["Review Rating"].mean().sort_values()

plt.barh(brand_ratings.index, brand_ratings.values)

plt.title("Average Customer Rating by Brand")
plt.xlabel("Average Rating")
plt.ylabel("Brand")

plt.xlim(0, 5)
plt.show()
```

```python
category_counts = df["Category"].value_counts()

plt.bar(category_counts.index, category_counts.values)

plt.title("Number of Customer Reviews by Category")
plt.xlabel("Category")
plt.ylabel("Number of Reviews")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
product_ratings = df.groupby("Asin")["Review Rating"].mean().sort_values()

plt.barh(product_ratings.index, product_ratings.values)

plt.title("Average Customer Rating by Product")
plt.xlabel("Average Rating")
plt.ylabel("Product ASIN")

plt.xlim(0, 5)
plt.tight_layout()
plt.show()
```

```python
complaint_counts = {
    "Skin irritation": 3,
    "Product effectiveness": 2,
    "Product texture": 2,
    "Product quality": 1
}

plt.barh(
    list(complaint_counts.keys()),
    list(complaint_counts.values())
)

plt.title("Common Customer Complaints")
plt.xlabel("Number of Mentions")
plt.ylabel("Complaint Theme")

plt.tight_layout()
plt.show()
```
